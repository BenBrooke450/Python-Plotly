```python
import matplotlib.pyplot as plt

# Sample data
x = [5, 7, 8, 10, 12, 14, 15, 18, 20]
y = [12, 14, 17, 20, 22, 24, 25, 27, 30]

# Create a scatter plot
plt.scatter(x, y, color='purple', marker='o')

# Add titles and labels
plt.title('Simple Scatter Plot Example')
plt.xlabel('X Values')
plt.ylabel('Y Values')

# Show the chart
plt.show()
```

A scatter plot technically needs two variables (an x and a y) — that’s why it’s called scatter in 2D space.
But yes, you can make a scatter plot from just one list by letting Matplotlib auto-generate the other axis (usually the index).

```python
import matplotlib.pyplot as plt

# One list of numbers
y = [5, 9, 3, 7, 2, 10, 4, 6, 8]

# Use index positions as x-values
x = range(len(y))

plt.scatter(x, y, color='teal', s=80)

plt.title('Scatter Plot from One List')
plt.xlabel('Index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```
